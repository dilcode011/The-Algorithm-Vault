```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
np.random.seed(42)
n_samples = 100

# Feature 1: Highly relevant
X1 = np.random.rand(n_samples, 1)
# Feature 2: Highly relevant
X2 = np.random.rand(n_samples, 1)
# Feature 3: Useless Noise (Random numbers)
X3 = np.random.rand(n_samples, 1)
# Feature 4: Useless Noise (Random numbers)
X4 = np.random.rand(n_samples, 1)

# Stack them into one matrix X
X = np.hstack([X1, X2, X3, X4])

# Formula: y = 3*X1 + 5*X2 + 0*X3 + 0*X4 + noise
# Ideally, model should learn that X3 and X4 have 0 influence.
y = 3*X1 + 5*X2 + np.random.randn(n_samples, 1)
```

```python

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

print("Data Shape:", X_train.shape)
print("True Coefficients: [3, 5, 0, 0]")
```

```python
# A. Linear Regression
lr = LinearRegression()
lr.fit(X_train, y_train)

# B. Ridge (L2)
ridge = Ridge(alpha=0.5)
ridge.fit(X_train, y_train)

# C. Lasso (L1)
lasso = Lasso(alpha=0.1) 
lasso.fit(X_train, y_train)

```

```python
# Create a DataFrame to compare results side-by-side
coef_df = pd.DataFrame({
    'Feature': ['X1 (Real)', 'X2 (Real)', 'X3 (Noise)', 'X4 (Noise)'],
    'True_Value': [3, 5, 0, 0],
    'Linear_Reg': lr.coef_.flatten(),
    'Ridge (L2)': ridge.coef_.flatten(),
    'Lasso (L1)': lasso.coef_
})

print("\n--- Coefficient Comparison ---")
print(coef_df.round(3))

```

```python
# VISUAL OBSERVATION:
# Linear Regression: Will assign non-zero weights to Noise (X3, X4) due to overfitting.
# Ridge: Will shrink Noise weights, but they will still be e.g., 0.1 or 0.2.
# Lasso: Should make X3 and X4 exactly 0.0.

```

```python
x = np.arange(4)
width = 0.25

plt.figure(figsize=(10, 6))
plt.bar(x - width, lr.coef_.flatten(), width, label='Linear')
plt.bar(x, ridge.coef_.flatten(), width, label='Ridge')
plt.bar(x + width, lasso.coef_, width, label='Lasso', color='red')

plt.xticks(x, ['X1', 'X2', 'X3 (Noise)', 'X4 (Noise)'])
plt.ylabel('Coefficient Value')
plt.title('Lasso vs Ridge: Handling Noise Features')
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.show()
```
